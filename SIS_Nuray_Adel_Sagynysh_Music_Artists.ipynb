```python
def get_wiki_value(page_name, field_name):
    url = "https://en.wikipedia.org/wiki/" + page_name

    response = requests.get(url, headers=headers, timeout=20)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")
    infobox = soup.find("table", class_="infobox")

    if infobox is None:
        raise ValueError("Infobox was not found on the page: " + page_name)
    for row in infobox.find_all("tr"):
        label = row.find("th", recursive=False)
        value = row.find("td", recursive=False)
        if label and value:
            if label.get_text(" ", strip=True) == field_name:
                for reference in value.select("sup.reference"):
                    reference.decompose()
                return value.get_text(" ", strip=True)

    raise ValueError("Field not found: " + field_name)
```

```python
def calculate_career_years(years_text):
    text = re.sub(r"\[[^\]]*\]", "", years_text).lower()
    pattern = r"(\d{4})\s*[-–—−]\s*(\d{4}|present)"
    periods = re.findall(pattern, text)
    remaining_text = re.sub(pattern, "", text)
    if not periods or re.search(r"\d{4}", remaining_text):
        raise ValueError("Years need to be checked: " + years_text)
    total_years = 0
    previous_end = 0
    for start, end in periods:
        start = int(start)
        end = current_year if end == "present" else int(end)
        if start < previous_end or end < start or end > current_year:
            raise ValueError("Incorrect intervals: " + years_text)
        total_years += end - start
        previous_end = end
    return total_years
```

```python
wiki_rows = []

for name, page, disco, kind in artists:
    years_text = None
    albums_text = None
    career_years = None
    albums_count = None
    try:
        years_text = get_wiki_value(page, "Years active")
        career_years = calculate_career_years(years_text)
    except (requests.RequestException, ValueError) as error:
        print(name, "- error with years:", error)
    try:
        albums_text = get_wiki_value(disco, "Studio albums")

        if not albums_text.isdigit():
            raise ValueError("Number of albums needs to be checked: " + albums_text)

        albums_count = int(albums_text)
    except (requests.RequestException, ValueError) as error:
        print(name, "- error with albums:", error)

    wiki_rows.append({
        "artist": name,
        "years_active": years_text,
        "career_years": career_years,
        "studio_albums": albums_text,
        "albums": albums_count
    })

df_wiki = pd.DataFrame(wiki_rows)
```
